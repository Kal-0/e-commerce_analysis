# Add-to-Cart Analysis ETL

## Purpose
Analyze product ordering position in cart (add_to_cart_order) to understand shopping patterns and identify high-priority products added first to cart.

## Input
* **Source:** `big_data.silver.order_products`

## Output
* **Target:** `big_data.gold.vw_add_to_cart_analysis`
* **Refresh:** Real-time (always reflects current Silver data)

## SQL Logic
1. GROUP BY add_to_cart_order position
2. COUNT total items per position
3. Calculate reorder rate per position
4. ORDER BY position and LIMIT to top 10

```python
%sql
-- Add-to-Cart Analysis View
-- Purpose: Analyze cart position patterns and reorder behavior by add-to-cart order

CREATE OR REPLACE VIEW big_data.gold.vw_add_to_cart_analysis AS
SELECT 
  add_to_cart_order,
  COUNT(*) AS total_items,
  ROUND(SUM(CASE WHEN reordered THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS reorder_rate
FROM big_data.silver.order_products
GROUP BY add_to_cart_order
ORDER BY add_to_cart_order
LIMIT 10;
```
