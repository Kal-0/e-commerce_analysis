# Gold Layer - Add-to-Cart Analysis View

## Purpose
Analyze product ordering position in cart (add_to_cart_order) to understand shopping patterns and identify high-priority products.

## Type
**SQL View** (lightweight, real-time)

## Input
* **Source:** `big_data.silver.order_products` (33.8M rows)

## Output
* **Target:** `big_data.gold.vw_add_to_cart_analysis`
* **Rows:** 10 (positions 1-10)
* **Refresh:** Real-time (always reflects current Silver data)

## Use Cases
* 🛒 Cart behavior analysis
* 🎯 Identify "must-have" products (added first)
* 📊 Understand shopping patterns

## Why View (not Table)?
* ✅ Result is very small (10 rows)
* ✅ Query is fast (simple groupBy)
* ✅ Always synchronized with Silver
* ✅ Zero storage overhead

## SQL Logic
1. GROUP BY add_to_cart_order
2. COUNT items and calculate reorder rate per position
3. LIMIT to top 10 positions

## Execution
Run all cells sequentially. Expected runtime: ~20 seconds.

```python
%sql
-- Add-to-Cart Analysis View

CREATE OR REPLACE VIEW big_data.gold.vw_add_to_cart_analysis AS
SELECT 
  add_to_cart_order,
  COUNT(*) AS total_items,
  ROUND(SUM(CASE WHEN reordered THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS reorder_rate
FROM big_data.silver.order_products
GROUP BY add_to_cart_order
ORDER BY add_to_cart_order
LIMIT 10;
```
